```python
# 퀴즈
# 리스트에서 평균을 구한 후 평균보다 큰 값만 출력하여라.

num_list = [100, 55, 50, 30, 25, 10, 67, 88, 45]
avg = __builtins__.sum(num_list)/len(num_list)
print(f"{num_list} 의 평균은? {avg:.2f} ")
# 평균값보다 크면 리스트의 아이템 출력
print("평균보다 큰 값은? ", end=" ")
cnt = 0
while (cnt < len(num_list)) :
    if (num_list[cnt] >= avg) :
       print(num_list[cnt], end = " ")
    cnt += 1
print()
print("=" * 30)
```

```python
# 입력받은 데이타를 리스트에 추가한다
# 입력데이타가 q 이거나 Q 이면 종료
# 빈 리스트 정의
word_list = []
# 무한루프 스타일로 반복문 정의
while 1:
    data = input("데이타 입력 ...(q나 Q이면 종료)").strip()
    # while 탈출 조건
    if (data.upper() == 'Q'):
        break
    else:
        word_list.append(data)
print()
print(word_list)
```

```python
# continue
#  : 제어문안에서 다음 단계의 명령문 수행
# 1~20 사이의 숫자중에서 3의 배수이거나 7의 배수이면 미출력
cnt = 0
while (cnt < 20):
    cnt += 1
    if (cnt % 3 == 0) or (cnt % 7 == 0) :
        continue
    else:
        print(cnt, end= " ")
print("continue 테스트 종료")
```

```python
# 퀴즈
# 빈 리스트를 생성하고 데이타를 입력받아 리스트에 추가한다.
# 리스트에 삽입되는 데이타는 숫자이어야 한다.
# 음수의 숫자도 입력 대상이다
# 데이타 총 입력 횟수는 10번이며 리스트의 길이가 5이면 입력을 중단한다.

numList = []
for i in range(10):
  if i <= 10:
    a = input("데이터를 입력하세요: ").strip()
    if a.isdigit() or((a[0]=="-")and a[1:].isdigit):
      numList.append(a)
      print("데이터가 추가되었습니다.")
  if len(numList) == 5:
    break
print(f'리스트 출력{numList}')
```

```python
# for문을 이용한 아이템 순환
# for 아이템 변수 in 문자열| 리스트| 튜풀:
#    명령어
# for 키변수 in 딕셔너리 :
#    명령어
# for 키변서, 값변수, in 딕셔너리.items():
#   명령문
# 리스트 순회
grade_list = [90, 25, 67, 45, 80]
for grade in grade_list:
  if (grade >= 60):
      print(f" {grade} 점 합격")
  else:
    print(f" {grade} 점 불합격")
print('리스트 순회 테스트 종료')
```

```python
# 딕셔너리 순회
word_dict = { 'a':'apple', 'b':'banana', 'c':'coffee'}
print('='*30)
for key in word_dict:
    print(f" {key} => {word_dict[key]}")

# 키와 값이 따로 각각 들어옮
print('='*30)
for key, value in word_dict.items():
    print(f" {key} => {value}")
# value를 이용하여 값을 불러올 수 있음
```

```python
# 퀴즈
# 딕셔너리 값에 'a' 글자가 있는 아이템만 표시하고 총 갯수를 추력하여라.

print('='*30)
word_dict = {'a': 'africa', 's': 'say', 'c': 'coffee', 'd': 'drama', 'y':'yes'}
tot = 0 # a 글자가 있는 아이템의 갯수 누적 변수
for k, v in word_dict.items():  # [(키, 값), ...]
    if 'a' in v: # 'a' 로 v = value에 'a'가 들어간 값 확인
        print(f" {k} => {v}")
        tot += 1
print(f' 총갯수는? {tot}')
```

```python
# 리스트 내포
# listname - [data for]
# list = [ data for in range() for in range()]

num_list = []
for i in range(1,11):
  num_list.append(i)
print(num_list)

# 리스트 내포를 통해 위 코드를 축약
num_list2 = [i for i in range(1,11,2)]
print(num_list2)

num_list1 = []
for i in range(1,10):
  num_list1.append((3*i)-1)
print(num_list1)
# 리스트포방식
num_list2 = [(3*i)-1 for i in range(1,10)]
print(num_list2)
```

```python
# 빈리스트 생성후 for 문을 이용해서 추가하는 방식
result_list1 = []
for i in range(1, 6):
    result_list1.append(f"{i}번")
print(result_list1)

# 빈리스트 생성후 for 문을 이용해서 추가하는 방식
result_list1 = []
for i in range(1, 6):
    result_list1.append(f"{i}번")
print(result_list1)
# ['1번', '2번', '3번', '4번', '5번']
# 리스트 내포 방식
result_list2 = [f"{i}번" for i in range(1, 6)]
print(result_list2)
# ['1번', '2번', '3번', '4번', '5번']
```

```python
guguList1 = []
for i in range(2,10):
  for j in range(1,10):
    guguList1.append(i*j)
print(guguList1)
# 내포 방식으로 변경
guguList2 = [i*j for i in range(2,10) for j in range(1,10)]
print(guguList2)
```
