```python
from string import printable
# 생성자 메서드가 있는 형태
# 사각형 클래스 선언
class SquareObj:
    # 생성자 메서드(__init__) 선언
    # 첫인자가 self, 값 전달 역할
    def __init__(self, width, height, color):
        self.width = width # .width = widht 같은 이름을 안써도 된다.
        self.height = height
        self.color = color
    # 일반 메소드 정의 > 속성 출력 용도
    def printInfo(self):
        print(f'가로 크기 : {self.width}')
        print(f'세로 크기 : {self.height}')
        print(f'색상 : {self.color}')
    # 사각형의 넓이를 출력하는 메소드 정의
    # 별도 인자가 있는 경우
    def printArea(self, mark):
        print(f'사각형의 넓이 {mark} {self.width * self.height}')

# 사각형 인스턴스 생성
s1 = SquareObj(10, 20, '파랑')
s2 = SquareObj(5, 5, '오렌지')
print(s1.width, s1.height, s1.color)
print(s2.width, s2.height, s2.color)
print('='*20)
# 일반 메소드 호출
# 인스턴스명.메소드명(값...)
s1.printInfo()
s1.printArea(':')
print('='*10)
s2.printInfo()
s2.printArea(':')
```

```python
# 퀴즈
# 계산기 만들기
class Calculator :
  # 생성자 메소드 정의 - 2개의 숫자 속성을 가짐
  def __init__(self,x,y):
     self.x = x
     self.y = y
  # 계산기 메소드 정의
  def op (self):
    print(f'더하기 : {self.x + self.y}')
    print(f'빼기 : {self.x - self.y}')
    print(f'곱하기 : {self.x * self.y}')
    print(f'나누기 : {self.x / self.y:.2f}')
number = Calculator(10, -6)
number.op()
```

```python
# 메소드에 별도의 인자가 있는 경우의 클래스
# Bread 클래스
# 속성 > 이름,가격,칼로리,재료.브랜드
# 메소드 > 정보 출력, 빵 갯수에 따른 가격
class Bread:
  def __init__(self,name,price,cal,material,brand):
    self.name = name
    self.price = price
    self.cal = cal
    self.material = material
    self.brand = brand
  def printInfo(self):
    print(f"종류 : {self.name}")
    print(f"가격 : {self.price}")
    print(f"칼로리 : {self.cal}")
    print(f"재료 : {self.material}")
    print(f"브랜드 : {self.brand}")
  def bread_data(self, num):
    print(f"{self.name} 를(을) {num}개 주문하셨습니다.")
    print(f"주문 가격은? {self.price} x {num} = {self.price * num}")

bread = Bread('모카빵', 4000, 600, ('밀가루','설탕','향신료','버터'),'파리바게뜨')
bread.printInfo()
print()

bread.bread_data(4)

bread1 = Bread('맘모스빵', 5500, 1200, ('밀가루','설탕','우유','버터'),'파리바게뜨')
bread1.printInfo()
print()

bread1.bread_data(12)
```

```python
# 퀴즈
# Cat 클래스르 만들고 객체화 한 후 메소드를 호출하여라
# 속성 : kind, name, age, gender, animal_kind(고양이)
# 메소드 : 속성출력(print_info), 동작 : run, sleep(어디서), eat(무엇을)

class Cat:
  def __init__(self,kind,name,age,gender,animal_kind):
     self.kind = kind
     self.name = name
     self.age = age
     self.gender = gender
     self.animal_kind = animal_kind
  def print_info (self):
    print(f'종류 : {self.kind}')
    print(f'이름 : {self.name}')
    print(f'나이 : {self.age}')
    print(f'성별 : {self.gender}')
    print()
  def run (self):
    print(f"{self.animal_kind} {self.name}가 달린다.")
  def sleep (self,sleep):
    print(f"{self.animal_kind} {self.name}가 {sleep}에서 잔다.")
  def eat(self,eat):
    print(f"{self.animal_kind} {self.name}가 {eat}을(를) 먹는다.")
  def action_print(self):
    print()
    print(f'{self.animal_kind} {self.name} 의 아침 일상')
    self.eat('물')
    self.eat('사료')
    self.run()
    self.eat('간식')
    self.sleep('쇼파')
    print()

cat1 = Cat('코캣', '덩치', 1, '남','고양이')
cat2 = Cat('러시안블루', '나비', 5, '여','고양이')
cat1.run()
cat2.run()
cat1.sleep('캣타워')
cat2.sleep('택배 박스')
cat1.eat('사료')
cat2.eat('춥스')
print()
cat1.print_info()

cat1.action_print()
cat2.action_print()
```

```python
# 퀴즈
# 도서 관리와 관련된 클래스 Library를 프로그래밍 하여라.
# - 전체 도서 수와 도서 제목을 관리하는 클래스 변수 이용
class Library:
  total_books = 0
  book_titles = []
  def __init__(self,title,author):
    self.title = title
    self.author = author
    Library.total_books += 1
    Library.book_titles.append(title)
  def show_total_books(self):
    print(f'총 도서 수 : {Library.total_books}')
  def show_all_titles(self):
    print(f'도서 제목은 : {", ".join(Library.book_titles)}')
book1 = Library("파이썬 자료구조", "김철수")
book2 = Library("알고리즘 개론", "이영희")
book3 = Library("데이터 과학 입문", "박민준")
book1.show_total_books()
book1.show_all_titles()
```
