```python
# 예외 처리

# try ... except : 발생한 오류를 메시지로 출력
# try ... except else ... finally 등등

mylist = [ 10, 45, 56]
mytxt = '도레미파솔라시'
try:
    print(12/3)
    print(mylist[0])
    print(int(mytxt))
    print(mylist[10])
    print(12/0)
except ZeroDivisionError as e:
    print('ZeroDivisionError 발생', e)
except IndexError as e:
    print('IndexError 발생', e)
except ValueError as e:
    print('ValueError 발생', e)
print('예외처리 테스트 종료 ')
```

```python
numList = [-42,  '21',  '십오',  39, -66, '구십삼', -38,  15,  18,  63]
for item in numList:
  try:
    print(abs(item))
  except Exception as e:
    pass

# pass를 이요하여 오류를 무시하고 출력

reusltList= []
for i in numList:
  try:
    reusltList.append(abs(i))
  except:
    pass
print(reusltList)
```

```python
# try ... except ... else ...

# try:
#     에러가 날것 같은 명령어
#     ...
# except Exception as e:
#     에러 처리 명령어
#     e 는 에러메세지 별칭
# else:
#     에러가 발생하지 않았을 경우 명령어

print('='*20)
numList = [-42, '21', '십오', 39, -66, '구십삼', -38, 15, 18, 63]
resultList = []
for num in numList:
    try:
        data = abs(num)
    except:
        print(f'에러발생 => {num}')
    else:
        resultList.append(data)
print(numList)
print(resultList)
```

```python
# 퀴즈
#입력된 글자가 숫자가 아니라면 에러 메세지 출력, 입력된 글자가 모두 숫자라면 두 수를 더한 후 출력한다.
def add_except():
  try:
    add_input = int(input("숫자 입력 : ").strip())
    add_input1 = int(input("숫자 입력 : ").strip())
  except:
    print(f'입력 데이터가 숫자가 아닙니다.')
  else:
    print(f"{add_input} + {add_input1} = {add_input + add_input1}")

add_except()
```

```python
# 퀴즈
# 문자열 데이타 리스트에서 숫자 데이타만 별도 정수 형태로 변경한 후
# 새로운 리스트에 추가하여 결과를 출력하는 함수를 프로그래밍 하시오
# try except 예외처리 구문을 이용한다.
# - 정수 형태로 변경할 수 없는 데이타의 경우 메시지를 띄우는 대신 아무런 작업도
# 수행하지 않도록 pass 키워드를 이용한다.

def make_int_list(txt_list):
  num_list = []
  for txt in txt_list:
    try:
      num = int(txt)
    except:
      pass
    else:
      num_list.append(num)
  return num_list

numbers_list = make_int_list(["10", "20", "abc", "40"])
print(f"변환된 숫자 리스트: {numbers_list} 길이 {len(numbers_list)}\n")

numbers_list = make_int_list(["327", "파이썬", "-100", "123-890", "0", "3.14"])
print(f"변환된 숫자 리스트: {numbers_list} 길이 {len(numbers_list)}\n")
```

```python
# try except else finall 문
# try :
#   에러가 날 것 같은 명령어
# except 에러코드 & Exception as e:
#     에러처리명령어
# else:
#   에러가 나지 않았을 때 명령어
# finally:
#   무조건 실행 명령

def divide_except(x,y):
  try:
    result = x/y
  except Exception as e:
    print('에러발생 > ', e)
  else:
    print(f'{x}나누기{y}는(은) {result}')
  finally:
    print('예외처리 테스트 종료')

divide_except(100,200)
divide_except('삼',10)
divide_except(10,0)
```

```python
# 퀴즈
# 회원의 이메일 주소와 나이 데이터를 함수로 전달시켜 이메일 형식과 정수 형태의
# 데이터가 맞다면 특정 메세지가 출력되도록 프로그래밍 하시오.
# - try except els finally 예외처리 구문을 이용한다.
# - 이메일 데이터의 유효성 검사는 index() 문자열 함수를 이용한다.
# - 에러 메세지는 as e 구문을 이용하여 출력한다.

def register_user(e_mail,age):
  print(f"---[회원 등록 시도: {e_mail}]---")
  try:
    e_mail.index("@") and int(age)
  except Exception as e:
    print('등록 실패 : 입력 데이터 형식이 올바르지 않습니다.')
    print(f"({e})")
  else:
    print(f"등록 성공! [이메일: {e_mail},나이 : {age}]")
  print("---[회원가입 절차 완료]---")
  print()

register_user("dream@python.com", "25")
register_user("wrong_email.com", "30")
register_user("test@test.com", "twenty")
```

```python
# 퀴즈
# n개의 단어를 입력받은 후 output.txt 파일로 저장되도록 함수를 정의하고 호출하여라.
# inputWriteFile(5, 'output/output.txt', 'utf-8') 함수 호출시 결과

def inputWriteFile(n, f, l):
  a = []
  for i in range(n):
    a.append(input(f"단어 입력: "))

  # 파일 저장 로직을 함수 내부로 이동
  with open(f, "w", encoding=l) as flie:
    for word in a:
      flie.write(word + '\n')

  print(f'입력된 단어 리스트는 {a}입니다.')
  print(f'{len(a)} 개의 단어가 모두 저장되었습니다.')

# 함수 호출
inputWriteFile(5, '/content/drive/MyDrive/직업훈련/data/test.txt', 'utf-8')
```

```python
# 퀴즈
# 'data/coding.txt' 문서를 읽고 코딩이라는 단어가 들어간 어구를 리스트화
# 한 후 총 갯수를 출력하는 함수를 코딩하여라.
def fileread(fileUrl,option,word):
  # 최대한 수정 없이, mode 위치에 'r'을 추가하여 인코딩 오류를 해결합니다.
  with open(fileUrl, 'r', encoding=option) as file:
    data = file.read()
    data_list = data.split()

    result_list = []
    for item in data_list:
      if word in item:
        result_list.append(item)

  print(f"{word}이라는 글자가 들어간 어구 출력")
  print(result_list)
  print(f"총 갯수는? {len(result_list)}")

fileread('/content/drive/MyDrive/직업훈련/data/coding.txt','utf-8','코딩')
```

```python
# csv 개요
# 쉼표로 분리된 데이터 csv, tsv, dat, xls, json, xml, html 등
# 관련 모듈 csv, pandas
# 관련 함수 DictReader, DicWriter, reader, writer
# 파일변수 = open(csvFileURL, 접근모드(w,r,a),encoding='cp949,utf-8')
# csv 변수 = csv.CSV관련함수(파일변수)
# 파일변수.close()
# with open(csvFileURL, 접근모드(w|r|a), encoding='cp949|utf-8') as 파일변수:
# csv변수 = csv.CSV관련함수(파일변수)
import csv
f = open('/content/drive/MyDrive/직업훈련/data/data.csv','r',encoding='cp949')

csv_data = csv.reader(f)
print(csv_data)
data_list = list(csv_data)
print(data_list)
print('='*30)
header_list = data_list[0]
item_list = header_list[1:]

for header in header_list:
  print(header, end='\t')
print()
for item in item_list:
  print(item, end='\t')
print()

for row in data_list[1:]:
  for item in row[1:]:
    print(item, end='\t')
  print()
for cls_txt, name, kor, eng, math, bio in data_list[1:]:
  print(cls_txt, name, kor, eng, math, bio)
  print(f'{cls_txt},{name},{kor},{eng},{math},{bio}')
  print()

f.close()
```
