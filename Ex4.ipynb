```python
# 퀴즈
# 문자열로 구성된 데이터가 초기형태로 삽입된 리스트와 빈 튜플을 선언한 후
# 주석문 형태를 참조하여 리스트의 요소를 이용하여 튜플에 요소를 추가하는 코드를 작성

# 출력 예시
# step0 : user_list = ['sky_blue', 'admin2026', 'p@ssword', 'membership']
# step1 : user_tuple = ()
# step2 : user_tuple = ('sky',)
# step3 : user_tuple = ('sky', '2026')
# step4 : user_tuple = ('sky', '2026', '@swr')
# step5 : user_tuple = ('membership', 'sky', '2026', '@swr')

user_list = [ 'sky_blue', 'admin2026', 'p@ssword', 'membership' ]
print(f' step0 : user_list = {user_list} ')
user_tuple = ()
print(f' step1 : user_tuple = {user_tuple} ')

user_tuple += (user_list[0][:3], )
print(f' step2 : user_tuple = {user_tuple} ')

user_tuple += (user_list[1][5:],)
print(f' step3 : user_tuple = {user_tuple} ')

user_tuple += (user_list[2][1::2], )
print(f' step4 : user_tuple = {user_tuple} ')

a= list(user_tuple)
a.insert(0,user_list[3])
user_tuple = tuple(a)
print(f' step5 : user_tuple = {user_tuple} ')
```

```python
# 집합 자료형
# 집합에 관련된 것들을 쉽게 처리하기 위해 만들어진 자료형 집합 연산 가능 (교집합, 합집합, 차집합, 여집합 가능)
# 중복 x 순서 x 인덱싱, 슬라이싱 x Set()을 이용하여 객체 선언

# 생성
s1 = {'a','b','c','d'}
s2 = set([1,2,3,4,2,2,100]) # 중복이 되지 않기 때문에 2가 하나로 출력 + 순서 상관 x
s3 = set('도레미파솔라시')
s4 = set(('야구','농구', '농구', '배구'))
s5 = set({'a':'apart', 'b':'banana', 'c':'cup'}) # key_name 형식으로 나오며 옆에 있는 아이템값이 나오지 않음
s6 = set()
print(s1,s2,s3,s4,s5,s6,type(s1),type(s2),type(s3),type(s4),type(s5),type(s6))
#
```

```python
# 1. Create :초기 집합 생성 완료
# - 북산고교 명단 : {'채치수', '서태웅', '송태섭', '강백호'}
a = set(['채치수', '서태웅', '송태섭', '강백호'])
print("북산고 명단 : ",a)
# --------------------------------------
# 2. Read : 집합 데이터 조회
# - 현재 등록된 총 선수 인원수 : 4 명
# --------------------------------------
print("현재 등록된 총 선수 인원수 : ",len(a),"명")
# 3. Update : 선수 추가 등록
# - 정대만 추가 후 명단 : {'채치수', '서태웅', '정대만', '송태섭', '강백호'}
# - 라이벌 추가 후 명단 : {'윤대협', '채치수', '서태웅', '정대만', '이정환', '송태섭', '강백호'}
# --------------------------------------
a.add('정대만')
b = set(['윤대협', '채치수', '서태웅', '정대만', '이정환', '송태섭', '강백호'])
print("정대만 추가 후 명단 : ",a)
print("라이벌 추가 후 명단 : ",b)
# 4. Delete : 명단에서 선수 제외
# - 윤대협 제외 후 최종 명단 : {'채치수', '서태웅', '정대만', '이정환', '송태섭', '강백호'}
b.remove('윤대협')
print("윤대협 제외 후 최종 명단 : ",b)
```

```python
# 집합 자료형
# 집합 연산자와 집합연산 함수
# |(합집합) , -(차집합), &(교집합) , ^(대칭차집합)
# union(), difference(), intersection(), symmetric_differnce()
set1 = {'최', '박', '선우', '김', '이'}
set2 = {'신', '장', '윤', '김', '이'}
print()
print(f" 합집합 => {set1 | set2}")
print(f" 차집합 => {set1 - set2}")
print(f" 교집합 => {set1 & set2}")
print(f" 대칭차집합 => {set1 ^ set2}")
```

```python
# 퀴즈
# 4개의 집합으로 정의된 OTT 서비스 데이터를 이용하여
# 집합 연산을 적용하여 결과를 조회하는 파이썬 프로그래밍 코드를 작성하여라

netflix_shows = {"오징어게임", "더글로리", "지옥", "킹덤", "스위트홈"}
tving_shows = {"더글로리", "삼시세끼", "지옥", "대탈출", "환승연애"}
disney_shows = {"무빙", "카지노", "지옥", "스타워즈", "어벤져스"}
# 친구가 꼭 보라고 강력 추천한 '인생작' 명단
recommended_shows = {"오징어게임", "더글로리", "무빙", "나의아저씨", "미스터션샤인"}
```

```python
# 플릭스 콘텐츠 명단 : {'지옥', '더글로리', '오징어게임', '스위트홈', '킹덤'}
print(f'플릭스 콘텐츠 명단 : {netflix_shows}')
# - 티빙 콘텐츠 명단     : {'지옥', '더글로리', '삼시세끼', '대탈출', '환승연애'}
print(f'티빙 콘텐츠명단 : {tving_shows}')
# - 디즈니플러스 명단    : {'지옥', '무빙', '카지노', '스타워즈', '어벤져스'}
print(f'디즈니플러스 명단 : {disney_shows}')
# - 친구의 강력 추천 명단 : {'무빙', '더글로리', '오징어게임', '나의아저씨', '미스터션샤인'}
print(f'친구의 강력 추천 명단 : {recommended_shows}')
# [OTT 스마트 추천 시스템 연산 결과]
# 1. 가성비 최고 추천작 (추천 & 넷플릭스 & 티빙) : {'더글로리'}
print(f'가성비 최고 추천작 (추천 & 넷플릭스 & 티빙) : {netflix_shows & tving_shows & recommended_shows}')
# 2. 모든 플랫폼 공통 콘텐츠 (넷플릭스 & 티빙 & 디즈니) : {'지옥'}
print(f'모든 플랫폼 공통 콘텐츠 (넷플릭스 & 티빙 & 디즈니) : {netflix_shows & tving_shows & disney_shows}')
# 3. 별도 개별 구매 필요 콘텐츠 (추천 -나머지 전체)   : {'나의아저씨', '미스터션샤인'}
print(f'별도 개별 구매 필요 콘텐츠 (추천 - 나머지 전체)   : {recommended_shows - netflix_shows - tving_shows - disney_shows}')
# 4. 현재 넷플릭스나 티빙 중 내가 시청 가능한 전체 콘텐츠 목록   : {'지옥', '더글로리', '
# 오징어게임', '스위트홈', '킹덤', '삼시세끼', '대탈출', '환승연애'}
print(f'현재 넷플릭스나 티빙 중 내가 시청 가능한 전체 콘텐츠 목록   : {netflix_shows | tving_shows}')
```
