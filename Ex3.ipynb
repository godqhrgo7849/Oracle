```python
# 아이템 삭제
# remove(값), pop(index), pop(),clear()
# del 리스트명[index]
# del 리스트명
CL = ['인천','서울','부산','제주','대구','대전','충주','청주','수원','부천','광명','김포','김해']
CL.remove('제주')
print(CL)
CL.pop(3)
print(CL)
CL.pop()
print(CL)
del CL[5]
print(CL)
print(CL)
CL.clear()
print(CL)
del CL
# print(CL) < 삭제로 인해 NameError 발생
```

```python
# 정렬
# sort(), reverse(), sorted()
num_list = [100, 20, -90, 10000, 4, 67]
print()
print(num_list)
print(sorted(num_list)) # 오름차순
print(num_list)
print(sorted(num_list, reverse=True)) # 내림차순
print(num_list[::-1]) # 역순
print(num_list)
num_list.sort() # 원본 리스트에 바로 반영
print(num_list)
```

```python
a = [] # 리스트를 루프 밖에서 초기화

# 퀴즈
# 데이터를 입력받은 후 리시트에 추가하고 결과를 출력하라.

i = 0
for i in range(5):
  if(i==0):
    b = input("좋아하는 음식은? ")
    a.append(b)
  elif(i==1):
    b = input("최근 본 영화는? ")
    a.append(b)
  elif(i==2):
    b = input("좋아하는 가수는?  ")
    a.append(b)
  elif(i==3):
    b = input("최근 여행지? ")
    a.append(b)
  elif(i==4):
    b = input("좋아하는 숫자?  ")
    a.append(b)
print("당신에 관한 리스트 : ",a)
```

```python
# 퀴즈
# 태어난 년도를 입력받은 후 태어난 년도에 대한 띠를 출력한다
# 태어난 년도를 12로 나눈후 나머지를 구한다. 나머지에 따라서 띠를 출력한다.
# 띠 = 태어난년도%12
# 원숭이, 닭, 개, 돼지, 쥐, 소, 범, 토끼, 용, 뱀, 말, 양 (0  ........  11)

a = ['원숭이', '닭', '개', '돼지', '쥐', '소', '범', '토끼', '용', '뱀', '말', '양']
# 리스트에 다 입력하는 것이 아닌 split을 이용하여 나눌 방식을 지정 후 리스트에 추가할 수 있다.
# a.split(", ") 등등
b = int(input("태어난 년도를 입력하세요 : "))
c = b % 12
print(b,"년도 출생은 ",a[c],"입니다")
```

```python
# 다중 리스트
# 중첩리스트 조회 => 리스트명[행인덱스][열인덱스]
grade_list = [
    ['고길동', '김영희', '이둘리'], # 학생이름
    [100, 80, 85], # 국어 점수
    [55, 70, 35], # 영어 점수
    [80, 80, 100] # 수학 점수
]
print(grade_list)
print(grade_list[0])
print(grade_list[0][0], grade_list[1][0], grade_list[2][0], grade_list[-1][0])
# 국어점수의 총점과 평균 출력
# sum(리스트)
print(f" 국어 : 총점 {sum(grade_list[1])} 평균 {sum(grade_list[1])/len(grade_list[1]):.2f}")
# 특정 학생의 총점과 평균 출력
total = grade_list[1][-1] + grade_list[2][-1] + grade_list[3][-1]
print(f" {grade_list[0][-1]} : 총점 {total} 평균 {total/3:.2f}")
```

```python
# 퀴즈
# 매장에 남아있는 음료수들의 재고를 파악하고,
# 각 음료수별 총 자산 가치(재고 수량 * 단가)를 자동으로 계산해 주는 프로그램을 작성
# - 데이터는 아래 구조의 중첩 리스트로 정의 한 후 사용
# - 문자열 포맷팅(%, format(),f")등을 이용하여 출력한다.

drink_inventory= [["콜라", 5, 1500],
  ["사이다", 3, 1200],
  ["보리차", 10, 2000],
  ["삼다수", 30, 800],
  ["캔커피", 7, 2500],
]
while True:
  drink = input("주문하실 음료수를 입력하세요 : ")
  if drink == "콜라":
    drink_inventory[0][1] -= 1
    print("콜라의 남은 재고 수량은 ",drink_inventory[0][1],"이고","잔액은 ",{drink_inventory[0][1] * drink_inventory[0][2]},"원 입니다.")
  elif drink == "사이다":
   drink_inventory[1][1] -= 1
   print("사이다의 남은 재고 수량은 ",drink_inventory[1][1],"이고","잔액은 ",{drink_inventory[1][1] * drink_inventory[1][2]},"원 입니다.")
  elif drink == "보리차":
    drink_inventory[2][1] -= 1
    print("보리차의 남은 재고 수량은 ",drink_inventory[2][1],"이고","잔액은 ",{drink_inventory[2][1] * drink_inventory[2][2]},"원 입니다.")
  elif drink == "삼다수":
   drink_inventory[3][1] -= 1
   print("삼다수의 남은 재고 수량은 ",drink_inventory[3][1],"이고","잔액은 ",{drink_inventory[3][1] * drink_inventory[3][2]},"원 입니다.")
  elif drink == "캔커피":
    drink_inventory[4][1] -= 1
    print("캔커피의 남은 재고 수량은 ",drink_inventory[4][1],"이고","잔액은 ",{drink_inventory[4][1] * drink_inventory[4][2]},"원 입니다.")
  else:
    print("주문하신 음료수는 재고가 없습니다.")
    break
```
